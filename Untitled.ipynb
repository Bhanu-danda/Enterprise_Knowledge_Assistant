# Clean the extracted text ( first run the fun defined cell then Run cell of fun calling)

```python
import re

def clean_text(text):
    # Normalize line endings
    text = text.replace("\r\n", "\n").replace("\r", "\n")

    # Remove trailing spaces from each line
    text = "\n".join(line.rstrip() for line in text.splitlines())

    # Replace tabs with spaces
    text = text.replace("\t", " ")

    # Remove excessive blank lines
    text = re.sub(r"\n{3,}", "\n\n", text)

    # Remove leading/trailing whitespace
    text = text.strip()

    return text
```

```python
cleaned_pages = []

for page in pages:
    cleaned = clean_text(page["text"])

    cleaned_pages.append({
        "page_number": page["page_number"],
        "text": cleaned,
        "source": "Reproduction.pdf"
    })

print("Pages cleaned:", len(cleaned_pages))
```

# Chunking

#### Create A Splitter first 
###### Try to keep each chunk around 800 characters.
###### Keep roughly 150 characters from the previous chunk in the next one.

```python
# create Chunks
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = []

for page in cleaned_pages:
    page_chunks = splitter.split_text(page["text"])

    for chunk_index, chunk_text in enumerate(page_chunks):
        chunks.append({
            "text": chunk_text,
            "metadata": {
                "source": page["source"],
                "page": page["page_number"],
                "chunk": chunk_index
            }
        })

print("Total chunks:", len(chunks))
```

# Embeddings

```python
# Load the embedding model
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "BAAI/bge-small-en-v1.5"
)

chunk_texts = [chunk["text"] for chunk in chunks]

embeddings = embedding_model.encode(
    chunk_texts,
    show_progress_bar=True
)

print("Embeddings shape:", embeddings.shape)
```

# Chunk → Embedding → Vector Store

# FAISS

```python
import faiss
```

```python
import faiss

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("Vector dimension:", dimension)
print("Vectors in index:", index.ntotal)
```

```python
# problem:- faiss only knows vectors,it doesn't know about our chunks ,So we maintain mapping b.w faiss id's and our chunks
```

# Testing our Retrieval

```python
# Let's ask our RAG system a question without an LLM yet.
#example: What is male reproductive system?
# First convert the question into an embedding.
```

```python
query = "What is the male reproductive system?"

query_embedding = embedding_model.encode(
    [query]
).astype("float32")

print(query_embedding.shape)
```

```python
# Now search FAISS:
```

```python
# suppose we need TOP-3
k = 3

distances, indices = index.search(query_embedding, k)

print("Indices:", indices)
print("Distances:", distances)
```

```python
# Convert FAISS results back to text
```

```python
for rank, idx in enumerate(indices[0], start=1):
    chunk = chunks[idx]

    print(f"\n{'=' * 70}")
    print(f"RESULT {rank}")
    print(f"FAISS ID: {idx}")
    print(f"Page: {chunk['metadata']['page']}")
    print(f"Distance: {distances[0][rank - 1]}")
    print("=" * 70)
    print(chunk["text"])
```

# Next step: Build the Context

```python
def build_context(results):
    context_parts = []

    for result in results:
        page = result["metadata"]["page"]
        text = result["text"]

        context_parts.append(
            f"[Page {page}]\n{text}"
        )

    return "\n\n".join(context_parts)
```

```python
results = []

for rank, idx in enumerate(indices[0], start=1):
    results.append({
        "rank": rank,
        "text": chunks[idx]["text"],
        "metadata": chunks[idx]["metadata"],
        "distance": float(distances[0][rank - 1])
    })

print("Number of results:", len(results))
```

# Next ONE step: actually build the context

```python
context = build_context(results)

print(context)
```

# 🧠 Build the Prompt for the LLM

```python
def build_prompt(query, context):
    prompt = f"""
You are a document question-answering assistant.

Answer the user's question using only the information
provided in the context below.

If the answer cannot be found in the context,
say that the information is not available in the document.

Do not invent or assume information.

CONTEXT:
{context}

QUESTION:
{query}

ANSWER:
"""
    return prompt
```

# Now We Have
#### query → "What is the male reproductive system?"
#### context → our Top-3 retrieved chunks

```python
prompt = build_prompt(query, context)

print(prompt)
```
