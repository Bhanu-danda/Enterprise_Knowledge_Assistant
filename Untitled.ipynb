# Extract ALL pages

```python
pages = []

for page_number, page in enumerate(doc, start=1):
    text = page.get_text()

    pages.append({
        "page_number": page_number,
        "text": text
    })

print("Total pages:", len(pages))
```

# repeated lines exist in your actual PDF

```python
from collections import Counter

all_lines = []

for page in pages:
    for line in page["text"].splitlines():
        line = line.strip()

        if line:
            all_lines.append(line)

line_counts = Counter(all_lines)

for line, count in line_counts.most_common(20):
    print(count, "→", repr(line))
```

# Clean the extracted text ( first run the fun defined cell then Run cell of fun calling)

```python
import re

def clean_text(text):
    # Normalize line endings
    text = text.replace("\r\n", "\n").replace("\r", "\n")

    # Remove trailing spaces from each line
    text = "\n".join(line.rstrip() for line in text.splitlines())

    # Replace tabs with spaces
    text = text.replace("\t", " ")

    # Remove excessive blank lines
    text = re.sub(r"\n{3,}", "\n\n", text)

    # Remove leading/trailing whitespace
    text = text.strip()

    return text
```

```python
cleaned_pages = []

for page in pages:
    cleaned = clean_text(page["text"])

    cleaned_pages.append({
        "page_number": page["page_number"],
        "text": cleaned,
        "source": "Reproduction.pdf"
    })

print("Pages cleaned:", len(cleaned_pages))
```

# compare before vs after

```python
page_number = 1

original = pages[page_number - 1]["text"]
cleaned = cleaned_pages[page_number - 1]["text"]

print("----- ORIGINAL -----")
print(original[:1000])

print("\n----- CLEANED -----")
print(cleaned[:1000])
```

```python
for original, cleaned in zip(pages, cleaned_pages):
    print(
        f"Page {original['page_number']}: "
        f"{len(original['text'])} → {len(cleaned['text'])}"
    )
```
