```python
import pymupdf
```

```python
Purpose

Allows Python to read our PDF.

PDF
 ↓
PyMuPDF
 ↓
Python
```

```python
pdf_path = r"data/Reproduction.pdf" 
```

```python
doc = pymupdf.open(pdf_path)
print("Number of pages:", len(doc))
```

```python
Inspect a particular page
```

```python
page = doc[2]

text = page.get_text()

print(text[:2000])
```

```python
Extract ALL pages
```

```python
pages = []

for page_number, page in enumerate(doc, start=1):
    text = page.get_text()

    pages.append({
        "page_number": page_number,
        "text": text
    })

print("Total pages:", len(pages))
```

```python
doc.close()
```

```python
Inspect extracted data
```

```python
print(pages[0].keys())
print(pages[0]["page_number"])
print(pages[0]["text"][:500])
```

```python
8️⃣ Inspect page sizes
```

```python
for page in pages[:15]:
    print(
        f"Page {page['page_number']}: "
        f"{len(page['text'])} characters"
    )
```
